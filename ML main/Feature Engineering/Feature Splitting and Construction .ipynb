```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
```

```python
df = pd.read_csv(r"C:\Skills\Machine Learning\ML main\Understanding Data\Titanic-Dataset.csv")
```

```python
df.head()
```

```python
df = df.dropna()
```

```python
df.shape
```

```python
x=df.iloc[:,1:]
y=df['Survived']
```

```python
cross_val_score(LogisticRegression(),x,y,cv=10).mean()
```

# Feature Construction

```python
# now making a new column family_size
x['Family_size'] = x['SibSp']+x['Parch']+1
```

```python
x = x.drop(['SibSp','Parch'],axis=1)
```

```python
# Now again applying cv score 
cross_val_score(LogisticRegression(),x,y,cv=6).mean()
```

```python
# you can see we slightly increase our accuracy 
```
