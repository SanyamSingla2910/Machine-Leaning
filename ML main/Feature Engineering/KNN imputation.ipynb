```python
import pandas as pd 
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv(r"C:\Skills\Machine Learning\ML main\Understanding Data\Titanic-Dataset.csv",usecols=['Age','Pclass','Fare','Survived'])
```

```python
df.head()
```

```python
from sklearn.impute import KNNImputer,SimpleImputer
```

```python
# seeing how many percentage of missing values are there 
df.isnull().mean()
```

```python
# doing train test split 
# we will compare the performance of simple mean imputer and our knn imputer 
x_train , x_test , y_train , y_test = train_test_split(df.iloc[:,1:],df['Survived'],test_size=0.25,random_state=25)
```

```python
x_train.head()
```

```python
# applying the knn imputer 

knn = KNNImputer(n_neighbors=5,weights='uniform')

x_train_trf = knn.fit_transform(x_train)
x_test_trf = knn.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()

lr.fit(x_train_trf,y_train)
pred = lr.predict(x_test_trf)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,pred)
```

```python
# as you can see we got a accuracy of 71 
# chekcing for Simple imputer 
si = SimpleImputer(strategy='most_frequent')
x_train_trf2 = si.fit_transform(x_train)
x_test_trf2 = si.transform(x_test)

lr2 = LogisticRegression()

lr2.fit(x_train_trf2,y_train)
pred2 = lr2.predict(x_test_trf2)
```

```python
accuracy_score(y_test,pred2)
```
