```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
from sklearn.model_selection import train_test_split
x_train ,x_test,y_train ,y_test = train_test_split(df.drop(columns='has_covid'),df.iloc[:,[-1]],test_size=0.2,random_state=0)
```

```python
# i will not apply the label encoder because we are learning the column transformer 
x_train.head()
```

# Using Column Transformer 

```python
from sklearn.compose import ColumnTransformer

transform = ColumnTransformer(transformers=[
    ('tr1',OneHotEncoder(sparse_output=False , drop='first'),['gender','city']),
    ('tr2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough'])
],remainder='passthrough')
```

```python
transform.fit_transform(x_train).shape
```

```python
transform.transform(x_test).shape
```

```python
# Now we can see we did multiple transformations on single column 
# Column Transformer can make our work a lot easier when we are doing feature transformation
```
