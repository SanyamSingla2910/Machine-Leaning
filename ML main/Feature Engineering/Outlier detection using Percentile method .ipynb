```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = pd.read_csv('weight-height.csv')
```

```python
df.head(2)
```

```python
df.shape
```

```python
# there are 10k rows we will apply method on height
sns.histplot(df['Height'],kde=True)
plt.show()
# very normally distributed 
```

```python
ub = df['Height'].quantile(0.99)
lb = df['Height'].quantile(0.01)
```

```python
print(ub,lb)
```

# Trimming 

```python
new_df = df[(df['Height']>=lb)&(df['Height']<=ub)]
sns.histplot(new_df['Height'],kde=True)
```

# Capping 

```python
df['Height'] = np.where(df['Height']<lb,lb,np.where(df['Height']>ub,ub,df['Height']))
```

```python
sns.histplot(df['Height'],kde=True)
```
