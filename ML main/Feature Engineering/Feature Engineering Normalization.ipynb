```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('wine_data.csv',usecols=[0,1,2])
df.columns=['class','alcohol','malic acid']
df.sample(5)
```

```python
# firstly we will see how the data is 
sns.scatterplot(data=df,x='alcohol',y='malic acid',hue='class')
```

```python
# now we will do the train test split 
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.iloc[:,1:],df['class'],test_size=0.25,random_state=0)
```

```python
# now we will apply normalization min max scaling and then see how the data became 
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaler.fit(x_train)
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# now we can convert back to pandas data frame 
x_train_scaled = pd.DataFrame(x_train_scaled,columns=x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled,columns=x_test.columns)
```

```python
# we can see here the min and max of the training input has became 0 and 1
x_train_scaled.describe()
```

```python
# now we will visualize the data 
# before the scaling scatterplot vs after 
fig , (ax1,ax2) = plt.subplots(1,2,figsize=(12,5))
ax1.scatter(x_train['alcohol'],x_train['malic acid'],c=y_train)
ax2.scatter(x_train_scaled['alcohol'],x_train_scaled['malic acid'],c=y_train)
plt.show()
```
