```python
# importing the libraries 
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
# getting only required features 
df = df.iloc[:,2:]
```

```python
df.head()
```

```python
df.shape
# we can see there are 400 rows 
```

```python
# now we will do the train test split 
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(df.iloc[:,0:2],df.iloc[:,-1],test_size=0.25,random_state=0)
```

```python
# performing the scaling 
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
scaler.fit(x_train)

x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# we can convert the numpy arrays into dataframe also 
x_train_scaled = pd.DataFrame(x_train_scaled,columns=x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled,columns=x_test.columns)
```

```python
# now we can visualize the scatter plot before and after scaling 
sns.scatterplot(x=x_train['Age'],y=x_train['EstimatedSalary'])
plt.title('Before')
plt.show()
```

```python
# after scaling 
sns.scatterplot(x=x_train_scaled['Age'],y=x_train_scaled['EstimatedSalary'])
plt.title('After')
plt.show()
```
