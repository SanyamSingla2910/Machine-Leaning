```python
import pickle
import numpy as np
```

```python
Sex = pickle.load(open('Sex.pkl','rb'))
Embarked = pickle.load(open('Embarked.pkl','rb'))
clf = pickle.load(open('Model.pkl','rb'))
```

```python
# creating a new input 
test_input = np.array([[2,'male',31.0,0,0,105,'S']])
```

```python
test_input_sex = test_input[:,1].reshape(1,1)
```

```python
test_input_sex = Sex.transform(test_input_sex)
```

```python
test_input_sex
```

```python
test_input_embarked = Embarked.transform(test_input[:,-1].reshape(1,1))
```

```python
test_input_embarked
```

```python
test_input_age = test_input[:,2].reshape(1,1)
```

```python
test_input_age
```

```python
test_input_new = np.concatenate((test_input[:,[0,3,4,5]],test_input_age,test_input_embarked,test_input_sex),axis = 1)
```

```python
test_input_new
```

```python
# Now we will predict the Sutvived 
clf.predict(test_input_new)
```
