# Univariate analysis

```python
# we can plot a count plot for the type to see how much movies or tv shows are there
sns.countplot(data= df,x='type')
plt.show()
# we can clearly see that there are more movies than tv shows 
```

```python
# to check the percentage we can do pie plot
plt.pie(df['type'].value_counts(),labels=['Movies','Tv shows'],autopct='%1.1f%%')
plt.title('Movies vs Tv shows percentage')
plt.show()
```

```python
# we can check if in which year most movies we released 
sns.histplot(df['release_year'],bins=20)
# we can see that later movies releases increased
```

```python
# we can see which rating movies are more 
sns.countplot(x=df['rating'])
plt.xticks(rotation=45)
plt.show()
# we can see tvma has most movies
```

# Bivariate Analysis

```python
# If we need to check in every type how many movies and tv shows are there 
sns.countplot(x=df['rating'],hue=df['type'])
plt.xticks(rotation = 45)
plt.show()
```

```python
sns.histplot(data=df, x='release_year', hue='type')
# we can see tv shows came later than the movies
```
