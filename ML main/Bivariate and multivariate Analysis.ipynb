```python
import pandas as pd
import seaborn as sns 
```

```python
# Importing the datasets we are gonna use for analysis using seaborn datasets 
tips = sns.load_dataset('tips')
titanic = pd.read_csv(r'C:\Skills\Machine Learning\ML main\Understanding Data\Titanic-Dataset.csv')
flight = sns.load_dataset('flights')
iris = sns.load_dataset('iris')
```

# scatterplot (Numerical vs Numerical)

```python
# here we can see that who gave more bill also give more tip 
sns.scatterplot(x='total_bill',y='tip',data=tips,hue='smoker',style='sex',size='size')
# we can use the hue function to check for multiple things to check like if it is male or female
# here we used scatter plot to do analysis of 5 collumns
```

```python
# doing scatter plot on different dataset
sns.scatterplot(data=titanic,x='Age',y='Fare',hue='Survived',style='Sex')
```

# Barplot(Numerical vs categorical)

```python
# here we check fare of different classes
sns.barplot(data=titanic,x='Pclass',y='Fare',hue='Sex')
```

# Boxplot (Num vs categorical)

```python
sns.boxplot(data=titanic,x='Sex',y='Age',hue='Survived')
```

# Displot

```python
sns.displot(data=titanic,x='Pclass',y='Age')
```

# Heatmap(Categorical vs Categorical)

```python
table = pd.crosstab(titanic['Pclass'], titanic['Survived'])
sns.heatmap(table)
```

```python
# Heatmap for Numerical 
sns.heatmap(titanic.corr(numeric_only=True))
```

```python
# ploting the cluster map 
sns.clustermap(pd.crosstab(tips['sex'],tips['smoker']))
pd.crosstab(tips['sex'],tips['smoker'])
```

# pairplot

```python
# gives the scatterplot of all the numerical collumns of the dataframe
sns.pairplot(iris,hue = 'species')
```

# line plot(Numerical vs numerical)

```python
sns.lineplot(data=flight,x='year',y='passengers')
```

```python
# this is just an example where we can use heatmap
flight.head()
sns.heatmap(flight.pivot_table(values='passengers',columns='year',index='month'))
```
