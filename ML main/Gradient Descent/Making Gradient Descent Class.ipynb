```python
# We made our own class 
# Now we will make a class which automatically calculates the m and b 
class GDR:
    def __init__(self,learning_rate,epochs):
        self.lr = learning_rate
        self.epoch = epochs
        self.m = 0
        self.b = 0
    def fit(self,x,y):
        for i in range (self.epoch):
            slopeb = -2 * np.sum(y-(self.m*x.ravel())-(self.b))
            slopem = -2 * np.sum((y-(self.m*x.ravel())-(self.b))*x.ravel())
            self.b = self.b - self.lr*slopeb
            self.m = self.m - self.lr*slopem
        print(self.b," , ",self.m)
```

```python
gd = GDR(0.001,20)
```

```python
gd.fit(x,y)
```
