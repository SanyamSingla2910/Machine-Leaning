```python
import pandas as pd 
import numpy as np 
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
```

```python
x,y = load_diabetes(return_X_y=True)
```

```python
# Doing Train Test Split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=69)
```

```python
# Now applying Ridge Regression and lasso together using Elastic Net Regression
from sklearn.metrics import r2_score
from sklearn.linear_model import ElasticNet,Ridge,Lasso
```

```python
# Only Ridge 
Lr = Ridge(alpha=0.1)
Lr.fit(x_train,y_train)
r2_score(y_test,Lr.predict(x_test))
```

```python
# Only Lasso
Lr = Lasso(alpha=0.1)
Lr.fit(x_train,y_train)
r2_score(y_test,Lr.predict(x_test))
```

```python
# Both 
Lr = ElasticNet(alpha=0.001,l1_ratio=0.9)
Lr.fit(x_train,y_train)
r2_score(y_test,Lr.predict(x_test))
```

```python
# you can tweak to optimize the model for a much better results 
```

```python
# Applying grid search cv for hyper parameter tuning 
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import GridSearchCV

model = Lasso()

param_grid = {
    "alpha": [0.00001, 0.0001, 0.001, 0.01, 0.1, 1, 10]
}

grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    scoring="r2",
    n_jobs=-1,
    cv=5
)

grid.fit(x_train, y_train)

print("Best Parameters:", grid.best_params_)
print("Best CV Score:", grid.best_score_)
```
