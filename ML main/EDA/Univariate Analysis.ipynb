```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Skills\Machine Learning\ML main\Understanding Data\Titanic-Dataset.csv")
```

```python
# to see how the data is 
df.head()
```

# Categorical Data 

### CountPlot

```python
# For example i need to find how many persons survived or not or to check the sex male or female
sns.countplot(x=df['Survived'])
sns.countplot(x=df['Sex'])
```

### Pie Chart

```python
plt.pie(df['Survived'].value_counts(),labels=['Not Survived','Survived'],autopct='%1.1f%%')
plt.title('Survived percentages')
plt.show()
```

```python
# This was all for the categorical data now lets move to numerical data 
```

# Numerical data

### Histogram

```python
# We can see that middle aged people were present more in the titanic
sns.histplot(x = df['Age'])
plt.show()
```

#### Distplot

```python
# This is a better version of histplot 
sns.displot(df['Age'],kde=True)
plt.show()
```

### Boxplot

```python
sns.boxplot(df['Fare'])
```
